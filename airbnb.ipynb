```python
missing_values = df.isnull().sum()
missing_percent = (df.isnull().mean() * 100).round(2)
missing_report = pd.concat([missing_values, missing_percent], axis=1)
missing_report.columns = ["missing_values", "missing_percentage"]
missing_report
```

```python
df["name"] = df["name"].fillna("Unknown")
df["host_name"] = df["host_name"].fillna("Unknown")
df["last_review"] = df["last_review"].fillna("Unknown")
df["reviews_per_month"] = df["reviews_per_month"].fillna(df["reviews_per_month"].median())
df["reviews_per_month"] = df["reviews_per_month"].fillna(0)
df.isnull().sum()
```

```python
print("Duplicate rows:", df.duplicated().sum())
df = df.drop_duplicates().reset_index(drop=True)
df = df[(df["price"] > 0) & (df["minimum_nights"] > 0) & (df["minimum_nights"] <= 365)].copy()
print("Rows after cleaning:", df.shape[0])
```

```python
df["last_review"] = pd.to_datetime(df["last_review"], errors="coerce")
df["price"] = pd.to_numeric(df["price"], errors="coerce")
df["reviews_per_month"] = pd.to_numeric(df["reviews_per_month"], errors="coerce")
df["neighbourhood_group"] = df["neighbourhood_group"].astype(str)
df["room_type"] = df["room_type"].astype(str)
print(df.dtypes)
```

```python
df["price_log"] = np.log1p(df["price"])
df["price_category"] = pd.cut(df["price"], bins=[0, 100, 200, 500, np.inf], labels=["Budget", "Moderate", "Expensive", "Luxury"], right=False)
df["review_rate"] = df["number_of_reviews"] / (df["reviews_per_month"].replace(0, np.nan) + 1)
df["host_listing_group"] = np.where(df["calculated_host_listings_count"] > 1, "Multiple", "Single")
df[["price", "price_log", "price_category", "review_rate", "host_listing_group"]].head()
```

```python
avg_price_by_borough = df.groupby("neighbourhood_group")["price"].mean().sort_values(ascending=False)
avg_price_by_borough
```

```python
df.sample(10, random_state=42)
```

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df.select_dtypes(include=['object']).columns.tolist()
print('Numerical columns:', numeric_cols)
print('Categorical columns:', cat_cols)
print('Missing values:', df.isnull().sum().sum())
```

```python
df["neighbourhood_group"].value_counts()

plt.figure(figsize=(8,5))
sns.countplot(data=df, x="neighbourhood_group", order=df["neighbourhood_group"].value_counts().index)
plt.title("Airbnb Listings by Borough")
plt.xlabel("Neighbourhood Group")
plt.ylabel("Count")
plt.xticks(rotation=30)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df, x="neighbourhood_group", y="price")
plt.title("Price by Neighbourhood Group")
plt.xlabel("Borough")
plt.ylabel("Price")
plt.xticks(rotation=30)
plt.show()

plt.figure(figsize=(8,5))
sns.histplot(data=df, x="price", bins=40, kde=True)
plt.title("Distribution of Airbnb Prices")
plt.xlabel("Price")
plt.ylabel("Count")
plt.show()
```

```python
room_prices = df.groupby("room_type")["price"].mean().sort_values(ascending=False)
room_prices

plt.figure(figsize=(8,5))
sns.barplot(x=room_prices.index, y=room_prices.values)
plt.title("Average Price by Room Type")
plt.xlabel("Room Type")
plt.ylabel("Average Price")
plt.xticks(rotation=20)
plt.show()
```

```python
top_neighbourhoods = df["neighbourhood"].value_counts().head(10)
top_neighbourhoods

plt.figure(figsize=(10,6))
top_neighbourhoods.plot(kind="bar")
plt.title("Top 10 Neighbourhoods by Listings")
plt.xlabel("Neighbourhood")
plt.ylabel("Number of Listings")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='longitude', y='latitude', hue='neighbourhood_group', s=20, alpha=0.6)
plt.title('Airbnb Listing Locations by Borough')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.legend(title='Borough', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()

corr = df[["price", "minimum_nights", "number_of_reviews", "reviews_per_month", "calculated_host_listings_count", "availability_365"]].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation Heatmap')
plt.show()
```

```python
df.groupby("neighbourhood_group")["price"].agg(["mean", "min", "max"]).sort_values("mean", ascending=False)

expensive_neighbourhoods = df.groupby("neighbourhood")["price"].mean().sort_values(ascending=False).head(10)
expensive_neighbourhoods

plt.figure(figsize=(10,6))
expensive_neighbourhoods.plot(kind='bar')
plt.title('Top 10 Most Expensive Neighbourhoods')
plt.xlabel('Neighbourhood')
plt.ylabel('Average Price')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(9,6))
sns.scatterplot(data=df, x='number_of_reviews', y='price', alpha=0.5)
plt.title('Price vs Number of Reviews')
plt.xlabel('Number of Reviews')
plt.ylabel('Price')
plt.show()

plt.figure(figsize=(9,6))
sns.scatterplot(data=df, x='reviews_per_month', y='price', alpha=0.5)
plt.title('Price vs Reviews per Month')
plt.xlabel('Reviews per Month')
plt.ylabel('Price')
plt.show()
```
