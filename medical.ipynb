```python
missing = df.isnull().sum()
missing_pct = (df.isnull().mean() * 100).round(2)
missing_report = pd.concat([missing, missing_pct], axis=1, keys=['missing_values', 'missing_pct'])
missing_report[missing_report['missing_values'] > 0]
```

```python
df = df.rename(columns={'Hipertension':'Hypertension', 'No-show':'No_show'})
df.columns.tolist()
```

```python
df['ScheduledDay'] = pd.to_datetime(df['ScheduledDay'])
df['AppointmentDay'] = pd.to_datetime(df['AppointmentDay'])
df[['ScheduledDay', 'AppointmentDay']].head()
```

```python
df['Age'] = df['Age'].clip(lower=0)
print('Negative ages:', (df['Age'] < 0).sum())
print('Max age:', df['Age'].max())
```

```python
df['WaitingTimeDays'] = (df['AppointmentDay'] - df['ScheduledDay']).dt.days

df['AgeGroup'] = pd.cut(
    df['Age'],
    bins=[0, 12, 18, 35, 60, 100],
    labels=['Child', 'Teen', 'Young Adult', 'Adult', 'Senior'],
    include_lowest=True
)

df['No_show_num'] = df['No_show'].map({'No': 0, 'Yes': 1})
df['WaitingTimeCategory'] = pd.cut(
    df['WaitingTimeDays'],
    bins=[-1, 0, 7, 14, 30, 1000],
    labels=['Same Day', 'Within a Week', 'Within 2 Weeks', 'Within a Month', 'More than a Month'],
    include_lowest=True
)

df[['Age', 'AgeGroup', 'WaitingTimeDays', 'No_show', 'No_show_num']].head()
```

```python
overall_no_show = df['No_show_num'].mean() * 100
print('Overall no-show rate:', round(overall_no_show, 2), '%')

no_show_gender = df.groupby('Gender')['No_show_num'].mean().sort_values(ascending=False)
print('No-show by gender:\n', no_show_gender)

no_show_age = df.groupby('AgeGroup')['No_show_num'].mean().sort_values(ascending=False)
print('No-show by age group:\n', no_show_age)

no_show_sms = df.groupby('SMS_received')['No_show_num'].mean().sort_values(ascending=False)
print('No-show by SMS reminder:\n', no_show_sms)

no_show_scholarship = df.groupby('Scholarship')['No_show_num'].mean().sort_values(ascending=False)
print('No-show by scholarship:\n', no_show_scholarship)

medical_conditions = df[['Hypertension', 'Diabetes', 'Alcoholism', 'No_show_num']].groupby(['Hypertension', 'Diabetes', 'Alcoholism']).mean().reset_index()
print('No-show by medical conditions:\n', medical_conditions.head())

plt.figure(figsize=(12, 6))
plt.subplot(2, 3, 1)
sns.countplot(data=df, x='No_show', palette='Set2')
plt.title('Appointment Outcome')

plt.subplot(2, 3, 2)
sns.barplot(x=no_show_gender.index, y=no_show_gender.values, palette='viridis')
plt.title('No-show by Gender')
plt.ylabel('Rate')

plt.subplot(2, 3, 3)
sns.barplot(x=no_show_age.index, y=no_show_age.values, palette='rocket')
plt.title('No-show by Age Group')
plt.ylabel('Rate')

plt.subplot(2, 3, 4)
sns.barplot(x=no_show_sms.index, y=no_show_sms.values, palette='magma')
plt.title('No-show by SMS Reminder')
plt.ylabel('Rate')

plt.subplot(2, 3, 5)
sns.barplot(x=no_show_scholarship.index, y=no_show_scholarship.values, palette='deep')
plt.title('No-show by Scholarship')
plt.ylabel('Rate')

plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
waiting_rate = df.groupby('WaitingTimeCategory')['No_show_num'].mean().sort_values(ascending=False)
sns.barplot(x=waiting_rate.index, y=waiting_rate.values, palette='coolwarm')
plt.title('No-show by Waiting Time Category')
plt.ylabel('No-show Rate')
plt.xticks(rotation=20)
plt.show()

plt.figure(figsize=(8, 5))
sns.histplot(df['Age'], bins=20, kde=True)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()

num_cols = df.select_dtypes(include=['number']).columns.tolist()
correlation = df[num_cols].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()

summary = df.groupby('No_show').agg(
    avg_waiting_time=('WaitingTimeDays', 'mean'),
    avg_age=('Age', 'mean'),
    total_appointments=('AppointmentID', 'count')
).round(2)
print(summary)

print('Final rows:', df.shape[0])
print('Final columns:', df.shape[1])
```
