```python
missing = df.isnull().sum()
missing_pct = (df.isnull().mean() * 100).round(2)
missing_report = pd.concat([missing, missing_pct], axis=1, keys=["missing_values", "missing_pct"])
missing_report[missing_report["missing_values"] > 0]
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df = df.drop_duplicates().reset_index(drop=True)
print("Rows after removing duplicates:", df.shape[0])
```

```python
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
df[['date_added']].head()
```

```python
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month
df[['date_added', 'year_added', 'month_added']].head()
```

```python
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna(df['country'].mode()[0])
df['rating'] = df['rating'].fillna(df['rating'].mode()[0])
df['duration'] = df['duration'].fillna(df['duration'].mode()[0])

for col in ['type', 'title', 'director', 'cast', 'country', 'rating', 'duration', 'listed_in', 'description']:
    df[col] = df[col].apply(lambda x: x.strip() if isinstance(x, str) else x)

df['country'] = df['country'].str.replace(r'\s+', ' ', regex=True).str.strip()

df['director'] = df['director'].replace('Unknown', 'Unknown')
print(df.isnull().sum().sort_values(ascending=False))
```

```python
df['duration_num'] = df['duration'].str.extract(r'(\d+)').astype(float)
df['duration_type'] = np.where(df['duration'].str.contains('Season', case=False, na=False), 'Seasons', 'Minutes')

df[['duration', 'duration_num', 'duration_type']].head()
```

```python
df['listed_in'] = df['listed_in'].fillna('General')
df['genre'] = df['listed_in'].str.split(',').str[0].str.strip()

df[['listed_in', 'genre']].head()
```

```python
type_counts = df['type'].value_counts()
print(type_counts)

plt.figure(figsize=(6, 5))
sns.countplot(data=df, x='type', palette='Set2')
plt.title('Movies vs TV Shows')
plt.xlabel('Type')
plt.ylabel('Count')
plt.show()
```

```python
titles_by_year = df['year_added'].value_counts().sort_index()
print(titles_by_year.head())

plt.figure(figsize=(12, 5))
titles_by_year.plot(kind='line', marker='o')
plt.title('Netflix Titles Added by Year')
plt.xlabel('Year')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()
```

```python
top_countries = df['country'].value_counts().head(10)
print(top_countries)

plt.figure(figsize=(10, 6))
ax = sns.barplot(x=top_countries.values, y=top_countries.index, palette='viridis')
plt.title('Top 10 Countries Producing Netflix Content')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.show()
```

```python
top_genres = df['genre'].value_counts().head(10)
print(top_genres)

plt.figure(figsize=(10, 6))
ax = sns.barplot(x=top_genres.values, y=top_genres.index, palette='magma')
plt.title('Top 10 Genres on Netflix')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')
plt.show()
```

```python
movie_data = df[df['type'] == 'Movie']
plt.figure(figsize=(10, 6))
sns.histplot(movie_data['duration_num'], bins=30, kde=True)
plt.title('Movie Duration Distribution')
plt.xlabel('Duration (minutes)')
plt.ylabel('Count')
plt.show()
```

```python
release_year_counts = df['release_year'].value_counts().sort_index()
print(release_year_counts.head())

plt.figure(figsize=(12, 5))
release_year_counts.plot(kind='line', marker='o', color='darkorange')
plt.title('Netflix Content by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
year_type = df.groupby(['release_year', 'type']).size().unstack(fill_value=0)
year_type.head()

plt.figure(figsize=(12, 6))
for col in year_type.columns:
    plt.plot(year_type.index, year_type[col], label=col)
plt.title('Movies vs TV Shows by Year')
plt.xlabel('Release Year')
plt.ylabel('Count')
plt.legend()
plt.show()
```

```python
rating_counts = df['rating'].value_counts().head(10)
print(rating_counts)

plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='rating', order=rating_counts.index, palette='deep')
plt.title('Netflix Ratings Distribution')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
num_cols = df.select_dtypes(include=['number']).columns.tolist()
correlation = df[num_cols].corr()
print(correlation)

plt.figure(figsize=(10, 8))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

sns.countplot(data=df, x='type', ax=axes[0,0], palette='Set2')
axes[0,0].set_title('Movies vs TV Shows')

year_type.plot(ax=axes[0,1])
axes[0,1].set_title('Content by Release Year')

sns.barplot(x=top_countries.values, y=top_countries.index, ax=axes[0,2], palette='viridis')
axes[0,2].set_title('Top Countries')

sns.histplot(movie_data['duration_num'], bins=25, kde=True, ax=axes[1,0])
axes[1,0].set_title('Movie Duration')

sns.countplot(data=df, x='rating', order=rating_counts.index, ax=axes[1,1])
axes[1,1].set_title('Ratings')

sns.barplot(x=top_genres.values, y=top_genres.index, ax=axes[1,2], palette='magma')
axes[1,2].set_title('Top Genres')

plt.tight_layout()
plt.show()
```

```python
insights = {
    'Most common type': df['type'].value_counts().idxmax(),
    'Most common country': df['country'].value_counts().idxmax(),
    'Most common genre': df['genre'].value_counts().idxmax(),
    'Peak release year': df['release_year'].value_counts().idxmax(),
    'Average movie duration': round(movie_data['duration_num'].mean(), 2),
}

for key, value in insights.items():
    print(f'{key}: {value}')

print('\nFinal cleaned dataset shape:', df.shape)
```
