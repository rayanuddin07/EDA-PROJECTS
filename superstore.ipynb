```python
print("Duplicate rows:", data.duplicated().sum())
print("Missing values:")
print(data.isnull().sum())
```

```python
data["Order Date"] = pd.to_datetime(data["Order Date"], format="%d/%m/%Y")
data["Ship Date"] = pd.to_datetime(data["Ship Date"], format="%d/%m/%Y")
print("Data types converted to datetime")
print(data.dtypes)
```

```python
print("Data Quality Check:")
print("Negative or zero sales:", (data["Sales"] <= 0).sum())
print("Sales range:", data["Sales"].min(), "to", data["Sales"].max())
print("\nUnique values:")
print("Segments:", data["Segment"].unique())
print("Regions:", data["Region"].unique())
print("Categories:", data["Category"].unique())
```

```python
data["Order_Year"] = data["Order Date"].dt.year
data["Order_Month"] = data["Order Date"].dt.month
data["Order_Quarter"] = data["Order Date"].dt.quarter
data["Shipping_Days"] = (data["Ship Date"] - data["Order Date"]).dt.days
print("Feature Engineering - Transformations Created:")
print(data[["Order_Year", "Order_Month", "Order_Quarter", "Shipping_Days"]].head(10))
print("\nShipping Days Statistics:")
print(data["Shipping_Days"].describe())
```

```python
print("BUSINESS METRICS")
print("\nTotal Sales:", data["Sales"].sum().round(2))
print("Average Sales per Order:", data["Sales"].mean().round(2))
print("Total Orders:", len(data))
print("Average Shipping Days:", data["Shipping_Days"].mean().round(2))

print("\nSALES BY CATEGORY")
category_sales = data.groupby("Category").agg({
    "Sales": ["sum", "mean", "count"]
}).round(2).sort_values(("Sales", "sum"), ascending=False)
print(category_sales)

print("\n TOP 10 SUB-CATEGORIES BY SALES ")
top_products = data.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False).head(10)
print(top_products.round(2))

print("\n SALES BY REGION")
region_sales = data.groupby("Region")["Sales"].sum().sort_values(ascending=False)
print(region_sales.round(2))

print("\nSALES BY SEGMENT")
segment_sales = data.groupby("Segment")["Sales"].sum().sort_values(ascending=False)
print(segment_sales.round(2))
```

```python
data.isnull().sum()
```

```python
print("CATEGORICAL DATA QUALITY")
for col in ["Ship Mode", "Segment", "Region", "Category", "Sub-Category"]:
    print(f"\n{col}: {data[col].nunique()} unique values")
    print(data[col].value_counts().head())
    
print("\nORDER ID DUPLICATES")
print("Total orders:", len(data))
print("Unique Order IDs:", data["Order ID"].nunique())
print("Duplicate Order IDs:", len(data) - data["Order ID"].nunique())
```

```python
print("SALES CATEGORIES TRANSFORMATION")
bins = [0, 100, 300, 25000]
labels = ["Low", "Medium", "High"]
data["Sales_Category"] = pd.cut(data["Sales"], bins=bins, labels=labels)
print(data["Sales_Category"].value_counts().sort_index())
print("\nSample:")
print(data[["Sales", "Sales_Category"]].head(10))
```

```python
print("OUTLIER DETECTION")
Q1 = data["Sales"].quantile(0.25)
Q3 = data["Sales"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[(data["Sales"] < lower_bound) | (data["Sales"] > upper_bound)]
print(f"Lower bound: {lower_bound:.2f}")
print(f"Upper bound: {upper_bound:.2f}")
print(f"Total outliers: {len(outliers)} ({len(outliers)/len(data)*100:.2f}%)")
print(f"\nOutlier Statistics:")
print(f"Min outlier: {outliers['Sales'].min():.2f}")
print(f"Max outlier: {outliers['Sales'].max():.2f}")
print(f"Mean outlier: {outliers['Sales'].mean():.2f}")
```

```python
print( "MONTHLY SALES TRENDS")
monthly_sales = data.groupby([data["Order Date"].dt.to_period("M")])["Sales"].agg(["sum", "mean", "count"]).round(2)
print(monthly_sales.tail(12))

print("\nYEARLY SALES SUMMARY")
yearly_sales = data.groupby("Order_Year").agg({
    "Sales": ["sum", "mean", "count"],
    "Shipping_Days": "mean"
}).round(2)
print(yearly_sales)

print("\nSALES TREND ANALYSIS")
sales_by_month = data.groupby(data["Order Date"].dt.to_period("M"))["Sales"].sum()
print(f"Min monthly sales: {sales_by_month.min():.2f}")
print(f"Max monthly sales: {sales_by_month.max():.2f}")
print(f"Average monthly sales: {sales_by_month.mean():.2f}")
```

```python
print(" COMPREHENSIVE ANALYSIS SUMMARY")

fig, axes = plt.subplots(2, 2, figsize=(16, 12))
monthly_data = data.groupby(data["Order Date"].dt.to_period("M"))["Sales"].sum()
monthly_data.plot(ax=axes[0, 0], marker='o', linewidth=2, color='steelblue')
axes[0, 0].set_title("Monthly Sales Trend", fontsize=13, fontweight="bold")
axes[0, 0].set_ylabel("Sales")
axes[0, 0].grid(True, alpha=0.3)
sales_pivot = data.pivot_table(values="Sales", index="Category", columns="Region", aggfunc="sum")
sns.heatmap(sales_pivot, annot=True, fmt=".0f", cmap="YlGn", ax=axes[0, 1], cbar_kws={'label': 'Sales'})
axes[0, 1].set_title("Sales by Category & Region", fontsize=13, fontweight="bold")

axes[1, 0].hist(data["Sales"], bins=50, color='skyblue', edgecolor='black', alpha=0.7)
axes[1, 0].axvline(upper_bound, color='red', linestyle='--', linewidth=2, label=f'Upper Bound: {upper_bound:.2f}')
axes[1, 0].axvline(lower_bound, color='orange', linestyle='--', linewidth=2, label=f'Lower Bound: {lower_bound:.2f}')
axes[1, 0].set_title("Sales Distribution with Outlier Bounds", fontsize=13, fontweight="bold")
axes[1, 0].set_xlabel("Sales")
axes[1, 0].set_ylabel("Frequency")
axes[1, 0].legend()

segment_shipping = data.groupby("Segment").agg({"Sales": "sum", "Shipping_Days": "mean"})
ax_twin = axes[1, 1].twinx()
segment_shipping["Sales"].plot(kind='bar', ax=axes[1, 1], color='steelblue', alpha=0.7, label='Total Sales')
segment_shipping["Shipping_Days"].plot(ax=ax_twin, marker='o', color='red', linewidth=2, markersize=8, label='Avg Shipping Days')
axes[1, 1].set_title("Sales & Shipping Days by Segment", fontsize=13, fontweight="bold")
axes[1, 1].set_ylabel("Total Sales", color='steelblue')
ax_twin.set_ylabel("Avg Shipping Days", color='red')
axes[1, 1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

print("\n✅ COMPLETE SUPERSTORE ANALYSIS FINISHED!")
print("\nFinal Dataset Summary:")
print(f"Total Rows: {len(data)}")
print(f"Total Columns: {len(data.columns)}")
print(f"Date Range: {data['Order Date'].min().date()} to {data['Order Date'].max().date()}")
print(f"Total Sales: ${data['Sales'].sum():,.2f}")
```
