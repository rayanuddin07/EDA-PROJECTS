```python
missing_percent = data.isnull().mean() * 100
missing_percent
```

```python
print("Duplicate rows:", data.duplicated().sum())
data[data.duplicated()].head()
```

```python
data["Age"] = data["Age"].fillna(data["Age"].median())
data["Embarked"] = data["Embarked"].fillna(data["Embarked"].mode()[0])
print("Missing values after filling:")
print(data.isnull().sum())
```

```python
data["HasCabin"] = data["Cabin"].notna().astype(int)
data.drop(columns=["Cabin"], inplace=True)
print("Cabin column converted and dropped")
```

```python
bins = [0, 12, 19, 60, 120]
labels = ["Child", "Teen", "Adult", "Senior"]
data["AgeGroup"] = pd.cut(data["Age"], bins=bins, labels=labels, right=False)
data[["Age", "AgeGroup"]].head(10)
```

```python
data["FamilySize"] = data["SibSp"] + data["Parch"] + 1
data["IsAlone"] = (data["FamilySize"] == 1).astype(int)
data[["SibSp", "Parch", "FamilySize", "IsAlone"]].head()
```

```python
bins = [0, 10, 30, 1000]
labels = ["Low", "Medium", "High"]
data["FareCategory"] = pd.cut(data["Fare"], bins=bins, labels=labels, right=False)
data[["Fare", "FareCategory"]].head(10)
```

```python
overall_survival_rate = data["Survived"].mean()
print("Overall survival rate:", round(overall_survival_rate * 100, 2), "%")
```

```python
print("Survival by Sex:")
print(data.groupby("Sex")["Survived"].mean().round(3))
print("\nSurvival by Pclass:")
print(data.groupby("Pclass")["Survived"].mean().round(3))
print("\nSurvival by AgeGroup:")
print(data.groupby("AgeGroup")["Survived"].mean().round(3))
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=data, x="Survived")
plt.title("Survival Count")
plt.xlabel("Survived")
plt.ylabel("Passengers")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=data, x="Pclass", y="Fare")
plt.title("Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=data, x="Pclass", y="Survived", hue="Sex")
plt.title("Survival Rate by Pclass and Sex")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=data, x="AgeGroup", hue="Survived")
plt.title("Survival by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Passengers")
plt.show()
```

```python
print("Survival by FamilySize:")
print(data.groupby("FamilySize")["Survived"].agg(["count", "mean"]).round(3))
print("\nSurvival by IsAlone:")
print(data.groupby("IsAlone")["Survived"].mean().round(3))
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=data, x="Embarked", hue="Survived")
plt.title("Survival by Port of Embarkation")
plt.xlabel("Embarked")
plt.ylabel("Passengers")
plt.show()
```
